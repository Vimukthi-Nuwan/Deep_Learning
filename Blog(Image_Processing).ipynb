# 🐱🐶 Cat vs Dog Image Classification using Deep Learning

## 📁 Dataset Overview  
This project focuses on classifying cat and dog images using various machine learning and deep learning models as part of my **Image Processing assignment**.

The dataset includes:
- **10,000 training images** (5,000 cats + 5,000 dogs)
- **1,000 test images** for prediction

All images were preprocessed (resized "224 * 224" and normalised) before training.

## 🧠 Model Experiments  

To explore different methodologies, I experimented with both traditional machine learning and modern deep learning models. Below is a summary of the validation accuracies I achieved using different approaches:

Screenshot 2025-03-14 at 10.41.41.png

🔍 **Observation**: Among all models, **VGG16** gave the highest validation accuracy and was selected as the final model.


## ✅ Final Model: VGG16

For my final implementation, I used the **VGG16** model with transfer learning. Key features of the model setup:

- **Pretrained weights** from ImageNet
- **Custom dense layers** with ReLU6 and sigmoid for binary classification
- **Dropout** to prevent overfitting
- **Adam optimizer** and **binary cross-entropy loss**

Training was performed over multiple epochs with batch normalisation and data augmentation to improve generalisation.
